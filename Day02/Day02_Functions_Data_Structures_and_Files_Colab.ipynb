## 2. Functions

```python
def add_gst(price, rate=18):
    """Return the price including GST."""
    tax = price * rate / 100
    return price + tax

print(add_gst(1000))
print(add_gst(1000, rate=5))
print(add_gst.__doc__)
```

```python
def total(*nums):
    return sum(nums)

def profile(**info):
    for key, value in info.items():
        print(f"{key} = {value}")

print(total(2, 4, 6))
profile(name="Arjun", city="Osaka", role="Engineer")

square = lambda x: x * x
print(square(5))
```

### Scope: local vs global

```python
count = 10          # global

def show():
    count = 99      # local, a different variable
    print("inside:", count)

show()
print("outside:", count)
```

## 3. Lists and list comprehensions

```python
fruits = ["apple", "mango", "kiwi"]
fruits.append("banana")
fruits.insert(1, "grape")
fruits.remove("kiwi")
print(fruits, fruits[0], fruits[-1], fruits[1:3], len(fruits))

prices = [120, 450, 80, 990]
with_tax = [round(p * 1.18, 2) for p in prices]
cheap = [p for p in prices if p < 200]
print(with_tax, cheap)
```

## 4. Tuples and sets

```python
point = (12.97, 77.59)
lat, lon = point          # unpacking
print(lat, lon)

emails = ["a@x.com", "b@x.com", "a@x.com"]
unique = set(emails)
print(unique, "a@x.com" in unique)

python_team = {"Ana", "Ken", "Ravi"}
ai_team = {"Ken", "Mei"}
print("Both:", python_team & ai_team)
print("Either:", python_team | ai_team)
```

## 5. Dictionaries and nested data

```python
student = {"name": "Yuki", "score": 88}
student["city"] = "Chennai"
print(student["name"], student.get("age", "N/A"))
for key, value in student.items():
    print(key, ":", value)

orders = [
    {"id": 1, "item": "Pen", "qty": 3, "price": 20},
    {"id": 2, "item": "Book", "qty": 1, "price": 350},
]
for o in orders:
    print(f'{o["item"]:<6} x{o["qty"]}  = {o["qty"] * o["price"]}')
print("Order value:", sum(o["qty"] * o["price"] for o in orders))
```

## 7. CSV files

```python
import csv
rows = [{"name": "Ana", "marks": 81}, {"name": "Ken", "marks": 67}, {"name": "Mei", "marks": 92}]
with open("marks.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "marks"])
    writer.writeheader()
    writer.writerows(rows)

with open("marks.csv") as f:
    data = list(csv.DictReader(f))
print(data)
print("Average:", sum(int(r["marks"]) for r in data) / len(data))
```

## 8. JSON files

```python
import json
config = {"app": "Planner", "version": 2, "tags": ["study", "python"], "active": True}
with open("config.json", "w") as f:
    json.dump(config, f, indent=2)

with open("config.json") as f:
    loaded = json.load(f)
print(loaded["tags"][0], type(loaded["version"]), loaded["active"])
print(json.dumps(loaded, indent=2))
```

## 10. Mini challenge: a contact book that remembers

```python
import json, os
FILE = "contacts.json"

def load_contacts():
    if not os.path.exists(FILE):
        return {}
    with open(FILE) as f:
        return json.load(f)

def save_contacts(contacts):
    with open(FILE, "w") as f:
        json.dump(contacts, f, indent=2)

def add_contact(contacts, name, phone):
    contacts[name] = phone

contacts = load_contacts()
add_contact(contacts, "Mei", "98450 11111")
add_contact(contacts, "Ravi", "99000 22222")
save_contacts(contacts)
print(load_contacts())
```
