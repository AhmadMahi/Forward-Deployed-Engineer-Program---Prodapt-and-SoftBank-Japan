## 2. Classes and objects

```python
class BankAccount:
    bank_name = "Sakura Bank"          # class attribute, shared

    def __init__(self, owner, balance=0):
        self.owner = owner              # instance attributes
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def __str__(self):
        return f"{self.owner}: {self.balance}"

a = BankAccount("Hana", 500)
b = BankAccount("Ravi")
a.deposit(250)
print(a, "|", b, "|", BankAccount.bank_name)
```

## 3. Inheritance and polymorphism

```python
class Account:
    def __init__(self, owner):
        self.owner = owner
    def fee(self):
        return 100

class Premium(Account):
    def fee(self):
        return super().fee() // 2

class Student(Account):
    def fee(self):
        return 0

for acc in [Account("A"), Premium("B"), Student("C")]:
    print(f"{type(acc).__name__:<8} fee = {acc.fee()}")
```

## 4. Properties and encapsulation

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price          # goes through the setter below

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("Price cannot be negative")
        self._price = value

    @property
    def price_with_tax(self):
        return round(self._price * 1.18, 2)

p = Product("Mouse", 500)
print(p.price, p.price_with_tax)
try:
    p.price = -10
except ValueError as e:
    print("Blocked:", e)
```

## 5. Exception handling

```python
def to_int(text):
    try:
        value = int(text)
    except ValueError:
        print(f"'{text}' is not a number")
        return None
    else:
        print("Converted OK")
        return value
    finally:
        print("Finished checking", repr(text))

print(to_int("42"))
print(to_int("abc"))
```

```python
class InsufficientFunds(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFunds(f"Balance {balance}, requested {amount}")
    return balance - amount

try:
    withdraw(100, 500)
except InsufficientFunds as e:
    print("Error:", e)
```

## 6. SOLID in action: open/closed with abstract classes

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount): ...

class UPI(Payment):
    def pay(self, amount):
        return f"Paid {amount} via UPI"

class Card(Payment):
    def pay(self, amount):
        return f"Paid {amount} by card"

def checkout(method: Payment, amount):
    print(method.pay(amount))      # depends on the abstraction, not a concrete class

checkout(UPI(), 499)
checkout(Card(), 1299)
```

## 8. NumPy

```python
import numpy as np
marks = np.array([72, 85, 90, 64, 78, 55, 92])
print("Plus 5:", marks + 5)
print("Mean:", marks.mean().round(2), "Max:", marks.max(), "Std:", marks.std().round(2))
print("Above 80:", marks[marks > 80])

grid = np.arange(1, 7).reshape(2, 3)
print(grid)
print("Column sums:", grid.sum(axis=0))
```

## 9. Decorators

```python
import time
from functools import wraps

def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        print(f"{func.__name__} took {time.perf_counter() - start:.4f} s")
        return result
    return wrapper

@timer
def slow_sum(n):
    return sum(range(n))

print(slow_sum(2_000_000))
```

## 10. Generators

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

print(list(countdown(5)))

import sys
as_list = [i * i for i in range(100_000)]
as_gen = (i * i for i in range(100_000))
print("List bytes:", sys.getsizeof(as_list), "| Generator bytes:", sys.getsizeof(as_gen))
```
