## 4. Variables and data types

````python
city = "Bengaluru"     # str
students = 30          # int
fee = 499.50           # float
is_open = True         # bool

print(city, students, fee, is_open)
print(type(city), type(students), type(fee), type(is_open))

students = students + 5    # variables can change
print("Students now:", students)
````

````python
# f-strings: a cleaner way to mix text and variables
print(f"{city} batch has {students} students. Fee: {fee:.2f}")
````

## 6. Conditions: `if`, `elif`, `else`

````python
marks = 72

if marks >= 85:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Keep practising")
````

````python
age, has_id = 20, True
print(5 == 5, 5 != 3, 10 > 3, 7 >= 7)
print("Can enter:", age >= 18 and has_id)
print("Weekend or holiday:", False or True)
print("Not raining:", not False)
````

## 7. Loops

### `for` loop with `range()`

````python
for n in range(1, 6):
    print("5 x", n, "=", 5 * n)

print(list(range(2, 11, 2)))   # start, stop (excluded), step
````

### `while` loop, `break` and `continue`

````python
attempts = 0
while attempts < 3:
    attempts += 1
    print("Attempt", attempts)
````

````python
for n in range(1, 11):
    if n == 4:
        continue      # skip 4
    if n == 8:
        break         # stop at 8
    print(n)
````

## 8. Putting it all together: a small grade checker

````python
# ===== Option A: interactive (active) =====
for i in range(3):
    score = int(input(f"Score {i + 1}: "))
    if score >= 50:
        print("Pass")
    else:
        print("Try again")

# ===== Option B: fixed test data, no typing needed (uncomment to use) =====
# for score in [72, 35, 50]:
#     print(score, "->", "Pass" if score >= 50 else "Try again")
````

## 9. Mini challenge: number guessing game

````python
# ===== Option A: fixed secret number for class demo (active) =====
secret = 13

# ===== Option B: random secret number (uncomment to use) =====
# import random
# secret = random.randint(1, 20)

attempts = 0
while attempts < 5:
    guess = int(input("Guess (1-20): "))
    attempts += 1
    if guess == secret:
        print(f"Correct! You got it in {attempts} attempts.")
        break
    elif guess < secret:
        print("Too low")
    else:
        print("Too high")
else:
    print("Out of attempts. The number was", secret)
````
